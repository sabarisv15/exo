```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

```python
dataset=pd.read_csv("./cumulativefiltered.csv")
```

```python
df3= dataset.fillna(dataset.mean(numeric_only=True))
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
df3['koi_disposition_encoded'] = label_encoder.fit_transform(df3['koi_disposition'])
```

```python
df3.head(5)
```

```python
x=df3.iloc[:,1:11]
y=df3.iloc[:,-1]
print(x.shape,y.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error as mse
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(
    n_estimators=600,     # number of trees
    max_depth=15,         # limit tree depth
    min_samples_split=5,  # prevent overfitting
    min_samples_leaf=3,   
    random_state=42,
    n_jobs=-1             # use all CPU cores
)
```

```python
model=RandomForestClassifier(random_state=50)
model.fit(x_train,y_train)
```

```python
pre=model.predict(x_test)
print(pre[:3])
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score
print("Accuracy:",accuracy_score(y_test,pre))
```

```python
from sklearn.model_selection import cross_val_score
```

```python
accuracy = accuracy_score(y_test,pre)
print("RandomForestClassifier Accuracy:", accuracy)
```
